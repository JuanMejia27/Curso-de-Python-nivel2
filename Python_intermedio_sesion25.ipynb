```python
import numpy as np

arr = np.array([1,2,3,4,5])
print("Array", arr)
print("Suma de elementos", arr.sum())

print("Promedio:", arr.mean())
```

```python
zeros = np.zeros(5)
ones = np.ones((3,3))
random = np.random.random(5)

print("Zeros", zeros)
print("Ones", ones)
print("Random", random)
```

```python
arr = np.arange(1,11,2)
print(arr)
```

```python
arr = np.arange(1,10).reshape(3,3)
print(arr)
```

```python
print(arr[0,1])
#Para imprimir por separado cada valor
```

```python
arr = np.array([2,4,6])
print(arr)
print(arr*2)
print(arr*3)
```

```python
arr1 = np.array([1,2,3])
arr2 = np.array([4,5,6])
print(arr1+arr2)
print(arr1*arr2)
```

```python
print(np.sqrt(arr1))
print(np.log(arr1))
print(np.sin(arr1))
```

```python
mat1 = np.array([[1,2],[3,4]])
mat2 = np.array([[5,6],[7,8]])

print(np.dot(mat1, mat2))
```

```python
data = np.array([10,20,30,40,50])
print(data.sum())
print(data.mean())
print(np.median(data))
print(np.std(data))
```

```python
print(data.max())
print(data.min())
print(data.argmax())
print(data.argmin())
```

```python
random_data = np.random.randint(1,101,size=10)
print(random_data)
print(random_data.mean())
```

```python
mat = np.array([[10,20],[30,40],[50,60]])
print(mat.sum(axis=1))
print(mat.sum(axis=0))
```

```python
#
data = np.random.randint(0,501,size=100)
print("datos originales\n", data)
```

```python
print(np.mean(data))
print(np.median(data))
print(np.std(data))
print("Maximo: ",np.max(data), "en", data.argmax())
print("Min:", np.min(data), "en", data.argmin())
```

```python
normalizacion = ((data - data.min()) / (data.max() - data.min()))
print(normalizacion)
```
