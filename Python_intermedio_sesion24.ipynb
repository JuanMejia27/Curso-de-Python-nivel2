```python
import pandas as pd

data = {
    "Nombre": ["Marian", "Lizeth", "Carlos", None],
    "Edad": [25, 27, None, 32],
    "Ciudad": ["Durango", "Guadalajara", None, "Tepic"]
}

df = pd.DataFrame(data)
print(df)
```

```python
df_sin_nulos = df.dropna()
print(df_sin_nulos)
```

```python
df_rellenado = df.fillna({"Nombre": "Desconocido", "Edad": round(df["Edad"].mean()), "Ciudad": "Sin ciudad"})
print(df_rellenado)
```

```python
data = {
    "Producto": ["Manzana", "Manzana", "Naranja", "Platano", "Naranja"],
    "Precio": [1.2,1.2,0.8,1.5,0.8]
}

df = pd.DataFrame(data)
print(df)
```

```python
print(df.duplicated())
```

```python
df_sin_duplicados = df.drop_duplicates()
print(df_sin_duplicados)
```

```python
data = {
    "Nombre": ["Marian", "Lizeth", "Carlos"],
    "Edad": [25, 27, 32],
    "Fecha": ["2023-01-01", "2023-01-02", "2023-01-03"]
}

df = pd.DataFrame(data)
print(df)
```

```python
df["Edad"] = df["Edad"].astype(float)
df["Fecha"] = pd.to_datetime(df["Fecha"])
print(df)
```

```python
df_renombrado = df.rename(columns={"Nombre": "Nombre completo", "Edad": "Años", "Fecha": "Tiempo"})
print(df_renombrado)
```

```python
df["Edad en meses"] = df["Edad"] * 12
print(df)
```

```python
data = {
    "Producto": ["Manzana", "Manzana", None, "Platano", "Naranja"],
    "Precio": [1.2,None,0.8,1.5,0.8],
    "Cantidad": [10,15,8,None,10]
}

df = pd.DataFrame(data)
print(df)
```

```python
df["Producto"] = df["Producto"].fillna("Desconocido")
df["Precio"] = df["Precio"].fillna(df["Precio"].mean())
df["Cantidad"] = df["Cantidad"].fillna(df["Cantidad"].mean())
df = df.drop_duplicates()

df["Product"] = df["Producto"].str.upper()
df["Total"] = df["Precio"] * df["Cantidad"]

print(df)
```
