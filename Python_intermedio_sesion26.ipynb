```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y, label = "Seno")
plt.title("Gráfico de lineas")
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
plt.legend()
plt.show()
```

```python
x = np.linspace(0, 10, 100)

plt.subplot(1, 2, 1)
plt.plot(x, np.sin(x), "r")
plt.title("Seno")

plt.subplot(1, 2, 2)
plt.plot(x, np.cos(x), "b")
plt.title("Coseno")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd

data = pd.DataFrame({
    "Categoria": ["A", "B", "C", "D"],
    "Valor": [8,7,15,5]
})

sns.barplot(x = "Categoria", y = "Valor", data = data)
plt.title("Grafico de barras con seaborn")
plt.show()
```

```python
data = np.random.normal(size = 100)
sns.distplot(data, kde=True, color="blue")
plt.title("Histograma con kde")
plt.show()
```

```python
tips = sns.load_dataset("tips")
sns.scatterplot(x = "total_bill", y = "tip", hue = "sex" , data = tips, palette = "cool")
plt.title("Relacion entre total de dinero y propinas")
plt.show()
```

```python
tips_encoded = pd.get_dummies(tips, drop_first=True)
correlation_matrix = tips_encoded.corr()
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt = ".2f")
plt.title("Matriz de correlacion")
plt.show()
```

```python
from ipywidgets import Label
sns.histplot(data, kde=True, color="blue")
plt.axvline(data.mean(), color="red", linestyle="--", label = "Promedio")
plt.legend()
plt.show()
```
