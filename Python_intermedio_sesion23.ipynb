```python
data = {
    "Nombre": ["Ana", "Esli", "Juan"],
    "Edad": [30,30,32],
    "ciudad": ["Guadalajara","Durango","Tepic"]
}

df = pd.DataFrame(data)
print(df)
```

```python
df.index = ["A", "B", "C"]
print(df)
```

```python
print(df["Nombre"])
```

```python
print(df[["Nombre", "Edad"]])
```

```python
filtro = df[df["Edad"] ==30]
print(filtro)
```

```python
df["Doble Edad"] = df["Edad"]*2
print(df)
```

```python
data = {
    "Pdoructo": ["Manzana", "Plantano","Sandia","Uva"],
    "Precio": [12,3,12,2],
    "Cantidad": [10,8,2,50]
}
df = pd.DataFrame(data)
print(df)
```

```python
df_filtrado = df[df["Precio"]>1]
print(df_filtrado)
```

```python
df["Total"] = df["Precio"] = df["Cantidad"]
print(df)
```
